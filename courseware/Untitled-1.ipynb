```python
# -*- coding: utf-8 -*-
"""
Spyder Editor

This is a temporary script file.
"""
import operator
class PolyList:
    def __init__(self):
        self.data=[]
        
    def Add(self,e):
        self.data.append(e)
        
    def CreateList(self,fname):
        fin=open(fname,"r")
        n=int(fin.readline().strip())
        for i in range(n):
            p=fin.readline().strip().split()
            self.data.append([float(p[0]),int(p[1])])
        fin.close()
        
        
    def getsize(self):
        return len(self.data)
    
    def __getitem__(self,i):
        return self.data[i]
    
    def getdata(self):
        return self.data
        print()
    
    def Sort(self):
        self.data=sorted(self.data,key=operator.itemgetter(1),reverse=True)
        
    def PolyAdd(self,B):
        C=PolyList()
        m=len(self.data)
        n=B.getsize()
        i,j=0,0
        while i<m and j<n:
            p,q=self.data[i],B[j]
            if p[1]>q[1]:
                C.Add(q)
                i+=1
            elif q[1]>p[1]:
                C.Add(q)
                j+=1
            else:
                k=p[0]+q[0]
                if(k!=0):
                    C.Add([k,p[1]])
                    i+=1
                    j+=1
        while i<m:
            p=self.data[i]
            C.Add(p)
            i+=1
        while j<n:
            q=B[j]
            C.Add(q)
            j+=1
        return C
```

```python
f=open('C:/Users/Ruijie Li/Desktop/代码/out.txt','w+')
p=PolyList()
p.CreateList('C:/Users/Ruijie Li/Desktop/代码/abc1.in.txt')

f.write('第一个多项式：')
p.getdata()
p.Sort()
f.write('排序后的结果：')
p.getdata()

q=PolyList()
q.CreateList('C:/Users/Ruijie Li/Desktop/代码/abc2.in.txt')
f.write('第二个多项式：')
q.getdata()
q.Sort()
f.write('排序后的结果：')
q.getdata()

r=p.PolyAdd(q)
f.write('相加多项式：')
r.getdata()
f.close()
```
