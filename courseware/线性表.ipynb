```python
%%ipythontutor height=600 width=1000

class ReportList:
	def __init__(self):			 					
		self.initcapacity=5       	  
		self.capacity=self.initcapacity
		self.name=[None]*self.capacity
		self.score=[None]*self.capacity
		self.size=0				
	def Add(self, name, socre): 			#在线性表的末尾添加一个元素e
		self.name[self.size]=name			#添加元素e
		self.score[self.size]=socre 
		self.size+=1
	def Resize(self, newcapacity):	
		assert newcapacity>=0		#检测参数正确性的断言
		oldname=self.name
		oldscore = self.score
		self.name=[None]*newcapacity
		self.score=[None]*newcapacity
		self.capacity=newcapacity
		for i in range(self.size):
			self.name[i]=oldname[i]
			self.score[i]=oldscore[i]
	def FindScore(self, name):
		for i in range(self.size):
			if self.name[i] == name:
				print("Name: %s, Score: %d" %(self.name[i],self.score[i]))
				break
			else:
				i = i+1
		if i == self.size:
			print("班级没有",name,"这个人")
	def Insert(self, i, name, score):
		assert 0<=i<=self.size        
		if self.size==self.capacity:
			self.Resize(2*self.size)
		for j in range(self.size,i-1,-1):
			self.name[j]=self.name[j-1]
			self.score[j]=self.score[j-1]
		self.name[i]=name      
		self.score[i]=score      
		self.size+=1
	def print_list(self):
		for i in range(self.size):
			print("Name: %s, Score: %d" %(self.name[i],self.score[i]))

    
	


    
```

```python
mylist = ReportList()
mylist.Add("张三", 90)
mylist.Add("李四", 85)
mylist.Add("王五", 92)
mylist.Add("赵六", 88)

mylist.FindScore("李四")
mylist.Insert(2, "孙七", 95)
mylist.print_list()
```

```python

class SqList:
    def __init__(self):
        self.initcapacity=5
        self.capacity=self.initcapacity
        self.data=[None]*self.capacity
        self.size=0
    def Add(self, e):
        if self.size==self.capacity:
            self.Resize(2*self.size)
        self.data[self.size]=e
        self.size+=1
    def Resize(self, newcapacity):
        assert newcapacity>=0
        olddata=self.data
        self.data=[None]*newcapacity
        self.capacity=newcapacity
        for i in range(self.size):
            self.data[i]=olddata[i]
    def getsize(self):
        return self.size
    def __getitem__(self, i):
        assert 0<=i<self.size
        return self.data[i]


def Merge2(A, B):                        	#求解算法
    C=SqList()                             	#新建顺序表C
    i=j=0                                  	#i用于遍历A,j用于遍历B
    while i<A.getsize() and j<B.getsize():	#两个表均没有遍历完毕
        if A[i]<B[j]:
            C.Add(A[i])                	#将较小的A[i]添加到C中
            i+=1
        else:
            C.Add(B[j])                	#将较小的B[j]添加到C中
            j+=1
    while i<A.getsize():			#若A没有遍历完毕
        C.Add(A[i])
        i+=1
    while j<B.getsize():			#若B没有遍历完毕
        C.Add(B[j])
        j+=1
    return C                               	#返回C
```

```python
def sum_of_natural_numbers(n):
    """
    计算自然数（1到n）的总和

    参数:
    n (int): 最大自然数

    返回:
    int: 自然数的总和
    """
    total = 0
    for i in range(1, n + 1):
        total += i
    return total

# 示例用法
n = 10
print("自然数 1 到", n, "的总和为:", sum_of_natural_numbers(n))
```
